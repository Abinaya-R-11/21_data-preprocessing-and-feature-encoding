## Stage 1 - Data Cleaning - drop, fillna with mean, median, mode Imputation, changing dtype

```python
weather_data.head(10)
```

```python
weather_data.drop(["Unnamed: 0","Temp C"],axis = 1, inplace= True)
```

```python
weather_data.head(15)
```

### I'm dropping this column, as 25% of the data is NaN.¶

```python
del weather_data["Ozone"]
```

```python
weather_data.isna().sum()
```

```python
weather_data[weather_data["Solar.R"].isna()]
```

```python
weather_data.groupby(by = "Weather")["Solar.R"].mean().round()
```

```python
weather_data["Solar.R"]=weather_data.groupby(by = "Weather")["Solar.R"].transform(lambda x:x.fillna(x.mean().round()))
weather_data
```

```python
weather_data.isna().sum()
```

```python
weather_data.dropna(inplace= True)
```

```python
weather_data.isna().sum()
```

```python
weather_data.dtypes
```

```python
weather_data["Month"].unique()
```

```python
weather_data["Month"].replace(to_replace="May",value="5",inplace=True)
```

```python
weather_data["Month"].unique()
```

```python
weather_data["Month"]=weather_data["Month"].astype(dtype=int)
weather_data
```

## Stage 2 - Data transformation

### 2. Data Transformation
* If data is CONTINUOUS = Standard Scaler, MinMaxScaler, Robust Scaler
* If data is DISCRETE   = Label Encoder, One hot Encoder

```python
weather_data_copy01 = weather_data.copy()
weather_data_copy01
```

```python
le_encoder = LabelEncoder()
weather_data_copy01["Weather"] = le_encoder.fit_transform(weather_data_copy01["Weather"])
weather_data_copy01
```

## 2. One Hot Encoder
This can be achieved by using any of the following 2 libraries:

1. Pandas - pd.get_dummies()
2. sklearn - OneHotEncoder()

```python
weather_data_copy02 = weather_data.copy()
weather_data_copy02
```

```python
weather_data_copy02= pd.get_dummies(data = weather_data_copy02,columns = ["Weather"],dtype = "int")
weather_data_copy02
```

```python
weather_data_copy03 = weather_data.copy()
weather_data_copy03
```

```python
ohe = OneHotEncoder()
weather_data_copy03 = ohe.fit_transform(weather_data_copy03["Weather"])
weather_data_copy03
```

### ====================================================================

### How to choose between Label Encoder and One Hot Encoder?
Input feature:

Based on whether it is Parametric or a Non Parametric Model, we will choosing between LE or OHE.


For Parametric Models - better to go with OHE.

For Non-Parametric Models - always go with Label Encoding.


Output feature:

Always the output feature, it must be Label Encoder.
